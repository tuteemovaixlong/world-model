```python
# Train MDN-RNN
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.distributions import Normal
from torch.utils.data import Dataset, DataLoader
import numpy as np
from tqdm import tqdm
import os

# ==========================================
# 1. DATASET SIÊU TỐC TỪ FILE .NPZ
# ==========================================
class LatentDataset(Dataset):
    def __init__(self, npz_path):
        print("Đang nạp toàn bộ Dữ liệu Latent vào RAM...")
        data = np.load(npz_path)
        self.z = data['z']              # Kích thước: (10001, 1000, 32)
        self.actions = data['actions']  # Kích thước: (10001, 1000, 3)
        print("✅ Nạp hoàn tất! Sẵn sàng huấn luyện tốc độ cao.")

    def __len__(self):
        return len(self.z)

    def __getitem__(self, idx):
        # Chuyển numpy thành tensor
        z_seq = torch.tensor(self.z[idx], dtype=torch.float32)
        a_seq = torch.tensor(self.actions[idx], dtype=torch.float32)

        # BÀI TOÁN DỰ ĐOÁN TƯƠNG LAI:
        # Input (z_t, a_t): từ bước 0 đến 998
        # Target (z_t+1): từ bước 1 đến 999
        z_in = z_seq[:-1]
        a_in = a_seq[:-1]
        z_target = z_seq[1:]

        return z_in, a_in, z_target

# ==========================================
# 2. KIẾN TRÚC MẠNG MDN-RNN CHUẨN MỰC
# ==========================================
class RNN(nn.Module):
    def __init__(self, latent_dim, action_dim, hidden_dim, num_layers=1):
        super().__init__()
        self.lstm = nn.LSTM(latent_dim + action_dim, hidden_dim, num_layers, batch_first=True)

    def forward(self, z, a, h=None):
        # Ghép nối đặc trưng ảnh (z) và hành động vô lăng (a)
        x = torch.cat([z, a], dim=-1)
        return self.lstm(x, h)

class MDN(nn.Module):
    def __init__(self, latent_dim, hidden_dim, num_gaussians):
        super().__init__()
        self.num_gaussians = num_gaussians
        self.latent_dim = latent_dim

        # 3 Nhánh output của mạng MDN
        self.fc_pi = nn.Linear(hidden_dim, num_gaussians)
        self.fc_mu = nn.Linear(hidden_dim, num_gaussians * latent_dim)
        self.fc_sigma = nn.Linear(hidden_dim, num_gaussians * latent_dim)

    def forward(self, outs_rnn, tau=1.0):
        pi = self.fc_pi(outs_rnn)
        mu = self.fc_mu(outs_rnn)
        sigma = self.fc_sigma(outs_rnn)

        batch_size, seq_length, _ = outs_rnn.size()

        # Định hình lại output cho khớp với số lượng Gaussian
        mu = mu.view(batch_size, seq_length, self.num_gaussians, self.latent_dim)
        sigma = sigma.view(batch_size, seq_length, self.num_gaussians, self.latent_dim)

        # Đảm bảo phương sai (sigma) luôn là số dương
        sigma = torch.exp(sigma) + 1e-8
        sigma = sigma * tau

        # Tính xác suất (trọng số) của các đỉnh Gaussian sao cho tổng = 1
        pi = F.softmax(pi / tau, dim=-1) + 1e-8

        return pi, mu, sigma

class MDNRNN(nn.Module):
    def __init__(self, latent_dim, action_dim, hidden_dim, num_gaussians, num_layers=1):
        super().__init__()
        self.rnn = RNN(latent_dim, action_dim, hidden_dim, num_layers)
        self.mdn = MDN(latent_dim, hidden_dim, num_gaussians)

    def forward(self, z, a, tau=1.0, h=None):
        outs_rnn, h = self.rnn(z, a, h)
        pi, mu, sigma = self.mdn(outs_rnn, tau)
        return pi, mu, sigma, h

# ==========================================
# 3. HÀM LOSS THEO CHUẨN TOÁN HỌC (NLL)
# ==========================================
def gaussian_nll_loss(pi, mu, sigma, z_target):
    z_target = z_target.unsqueeze(2)
    log_pi = torch.log(pi)

    # Tính xác suất phân bố chuẩn
    normal_dist = Normal(loc=mu, scale=sigma)
    log_prob = normal_dist.log_prob(z_target).sum(-1)

    # LogSumExp trick để tránh tràn số (Underflow/Overflow)
    log_sum_exp = torch.logsumexp(log_pi + log_prob, dim=-1)

    return -log_sum_exp.mean()

# ==========================================
# 4. KHAI HỎA HUẤN LUYỆN
# ==========================================
def train_rnn():
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print(f"🔥 Bắt đầu rèn luyện Memory Model (RNN) trên: {device}")

    # Đường dẫn file
    DATA_NPZ = '/content/drive/MyDrive/WorldModels/data/latent_data_10k.npz'
    CHECKPOINT_DIR = '/content/drive/MyDrive/WorldModels/checkpoints'
    os.makedirs(CHECKPOINT_DIR, exist_ok=True)

    # Cấu hình mạng (Chuẩn mực World Models)
    LATENT_DIM = 32
    ACTION_DIM = 3
    HIDDEN_DIM = 256
    NUM_GAUSSIAN = 5
    EPOCHS = 10         # Mạng RNN hội tụ khá nhanh
    BATCH_SIZE = 64     # Data siêu nhẹ -> Đẩy batch to để tận dụng VRAM

    dataset = LatentDataset(DATA_NPZ)
    dataloader = DataLoader(dataset, shuffle=True, batch_size=BATCH_SIZE)

    model = MDNRNN(LATENT_DIM, ACTION_DIM, HIDDEN_DIM, NUM_GAUSSIAN).to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

    for epoch in range(EPOCHS):
        model.train()
        total_loss = 0.0

        with tqdm(dataloader, desc=f"Epoch {epoch+1}/{EPOCHS}") as pbar:
            for z_in, a_in, z_target in pbar:
                z_in, a_in, z_target = z_in.to(device), a_in.to(device), z_target.to(device)

                optimizer.zero_grad()
                pi, mu, sigma, _ = model(z_in, a_in)

                loss = gaussian_nll_loss(pi, mu, sigma, z_target)

                loss.backward()
                # Kỹ thuật sinh tử của RNN: Cắt Gradient để chống "tràn" hoặc "nổ" số
                torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
                optimizer.step()

                total_loss += loss.item()
                pbar.set_postfix({'NLL Loss': f'{loss.item():.4f}'})

        avg_loss = total_loss / len(dataloader)
        print(f'✅ Epoch {epoch+1} | Avg NLL Loss: {avg_loss:.4f}')

        # Lưu checkpoint an toàn
        torch.save(model.state_dict(), f'{CHECKPOINT_DIR}/rnn_10k_epoch{epoch+1}.pth')

    # Chốt sổ bản final
    final_path = f'{CHECKPOINT_DIR}/rnn_10k_final.pth'
    torch.save(model.state_dict(), final_path)
    print(f"🎉 Đã lưu Memory Model thành công tại: {final_path}")

# Nhấn nút khởi động!
train_rnn()
```

## video visualization

```python
import gymnasium as gym
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML
import numpy as np
import torch
from torchvision import transforms
from tqdm import tqdm

def sample_mdn(pi, mu, sigma):
    component_index = torch.multinomial(pi.view(-1), 1).item()
    selected_mu = mu[0, 0, component_index]
    selected_sigma = sigma[0, 0, component_index]
    epsilon = torch.randn(selected_mu.size(), device=mu.device)
    z = selected_mu + selected_sigma * epsilon
    return z.unsqueeze(0).unsqueeze(0)

# Đã thêm tham số scale và tau vào định nghĩa hàm
def visualize_dream(env_name, vae, mdnrnn, transform, device, scale=5, tau=0.5, frames=300):
    env = gym.make(env_name, render_mode='rgb_array')
    obs, _ = env.reset()

    # Cho xe chạy thẳng lấy đà 50 step đầu tiên
    for _ in range(50):
        obs, _, _, _, _ = env.step(np.array([0.0, 1.0, 0.0]))

    mdnrnn.eval()
    vae.eval()

    h = (torch.zeros(1, 1, 256).to(device), torch.zeros(1, 1, 256).to(device))
    dream_frames = []

    print(f"Bắt đầu 'nằm mơ' sinh ra video với tau={tau}...")

    # Bắt đầu với frame thực tế cuối cùng
    obs_tensor = transform(obs).unsqueeze(0).to(device)
    with torch.no_grad():
        mu_vae, logvar = vae.encode(obs_tensor)
        z = vae.reparameterize(mu_vae, logvar).unsqueeze(0)

    for i in tqdm(range(frames)):
        action = np.array([0.0, 1.0, 0.0]) # Cứ tưởng tượng đang đạp ga đi thẳng
        action_tensor = torch.tensor(action, dtype=torch.float32).unsqueeze(0).unsqueeze(0).to(device)

        with torch.no_grad():
            # Sử dụng biến tau được truyền vào hàm
            pi, mu, sigma, h = mdnrnn(z, action_tensor, h=h, tau=tau)
            z_next = sample_mdn(pi, mu, sigma)
            reconstructed = vae.decode(z_next.squeeze(0))
            z = z_next

        img_np = (reconstructed.squeeze(0).permute(1, 2, 0).cpu().numpy() * 255).astype(np.uint8)
        dream_frames.append(img_np)

    env.close()

    # Xuất ra Video HTML (Dùng scale để điều chỉnh kích thước hiển thị)
    fig, ax = plt.subplots(figsize=(scale, scale))
    ax.axis('off')
    im = ax.imshow(dream_frames[0])

    def update(i):
        im.set_array(dream_frames[i])
        return [im]

    anim = animation.FuncAnimation(fig, update, frames=len(dream_frames), interval=50, blit=True)
    plt.close()
    return HTML(anim.to_html5_video())

# ==========================================
# THÊM ĐỊNH NGHĨA TRANSFORM VÀO ĐÂY
# ==========================================
transform = transforms.Compose([
    transforms.Lambda(lambda img: img[:-12, :, :]),
    transforms.ToPILImage(),
    transforms.Resize((96, 96), transforms.InterpolationMode.LANCZOS),
    transforms.ToTensor()
])

# Gọi hàm chạy thử (Lúc này hàm đã nhận params scale và tau)
visualize_dream("CarRacing-v3", vae_model, model, transform, device, scale=6, tau=0.01, frames=200)
```

```python
# Load VAE và MDN-RNN lên và train lại controller từ đầu
!pip install cma > /dev/null 2>&1

import cma
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import os
import time
import pandas as pd
import gymnasium as gym
from gymnasium.vector import AsyncVectorEnv

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
DRIVE_PATH = '/content/drive/MyDrive/WorldModels'
LATENT_DIM, ACTION_DIM, HIDDEN_DIM, NUM_GAUSSIANS = 32, 3, 256, 5

# ==========================================
# 1. ĐỊNH NGHĨA LẠI CÁC CLASS (VAE, RNN, CONTROLLER)
# ==========================================
class VAE(nn.Module):
    def __init__(self, image_channels=3, latent_dim=32):
        super(VAE, self).__init__()
        self.latent_dim = latent_dim
        self.encoder = nn.Sequential(
            nn.Conv2d(image_channels, 32, kernel_size=4, stride=2, padding=1), nn.ReLU(),
            nn.Conv2d(32, 64, kernel_size=4, stride=2, padding=1), nn.ReLU(),
            nn.Conv2d(64, 128, kernel_size=4, stride=2, padding=1), nn.ReLU(),
            nn.Conv2d(128, 256, kernel_size=4, stride=2, padding=1), nn.ReLU(),
            nn.Flatten()
        )
        self.fc_mu = nn.Linear(256 * 6 * 6, self.latent_dim)
        self.fc_logvar = nn.Linear(256 * 6 * 6, self.latent_dim)
        self.decoder_input = nn.Linear(latent_dim, 256 * 6 * 6)
        self.decoder = nn.Sequential(
            nn.Unflatten(1, (256, 6, 6)),
            nn.ConvTranspose2d(256, 128, kernel_size=4, stride=2, padding=1), nn.ReLU(),
            nn.ConvTranspose2d(128, 64, kernel_size=4, stride=2, padding=1), nn.ReLU(),
            nn.ConvTranspose2d(64, 32, kernel_size=4, stride=2, padding=1), nn.ReLU(),
            nn.ConvTranspose2d(32, image_channels, kernel_size=4, stride=2, padding=1),
            nn.Sigmoid()
        )

    def encode(self, x):
        h = self.encoder(x)
        return self.fc_mu(h), self.fc_logvar(h)

    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std

    def decode(self, z):
        return self.decoder(self.decoder_input(z))

    def forward(self, x):
        mu, logvar = self.encode(x)
        z = self.reparameterize(mu, logvar)
        return self.decode(z), mu, logvar

class RNN(nn.Module):
    def __init__(self, latent_dim, action_dim, hidden_dim, num_layers=1):
        super().__init__()
        self.lstm = nn.LSTM(latent_dim + action_dim, hidden_dim, num_layers, batch_first=True)
        self.hidden_dim = hidden_dim

    def forward(self, z, a, h=None):
        return self.lstm(torch.cat([z, a], dim=-1), h)

    def init_hidden(self, batch_size, device='cpu'):
        return (torch.zeros(1, batch_size, self.hidden_dim).to(device),
                torch.zeros(1, batch_size, self.hidden_dim).to(device))

class MDN(nn.Module):
    def __init__(self, latent_dim, hidden_dim, num_gaussians):
        super().__init__()
        self.num_gaussians = num_gaussians
        self.latent_dim = latent_dim
        self.fc_pi = nn.Linear(hidden_dim, num_gaussians)
        self.fc_mu = nn.Linear(hidden_dim, num_gaussians * latent_dim)
        self.fc_sigma = nn.Linear(hidden_dim, num_gaussians * latent_dim)

    def forward(self, outs_rnn, tau):
        tau = torch.tensor(tau).to(outs_rnn.device)
        pi = self.fc_pi(outs_rnn)
        mu = self.fc_mu(outs_rnn)
        sigma = self.fc_sigma(outs_rnn)
        batch_size, seq_length, _ = outs_rnn.size()
        mu = mu.view(batch_size, seq_length, self.num_gaussians, self.latent_dim)
        sigma = sigma.view(batch_size, seq_length, self.num_gaussians, self.latent_dim)
        sigma = torch.exp(sigma) + 1e-15
        sigma = sigma * torch.sqrt(tau)
        pi = F.softmax(pi/tau, dim=-1) + 1e-15
        return pi, mu, sigma

class MDNRNN(nn.Module):
    def __init__(self, latent_dim, action_dim, hidden_dim, num_gaussians, num_layers=1):
        super().__init__()
        self.rnn = RNN(latent_dim, action_dim, hidden_dim, num_layers)
        self.mdn = MDN(latent_dim, hidden_dim, num_gaussians)

    def forward(self, z, a, tau=1.0, h=None):
        outs_rnn, h = self.rnn(z, a, h)
        pi, mu, sigma = self.mdn(outs_rnn, tau)
        return pi, mu, sigma, h

class Controller(nn.Module):
    def __init__(self, state_dim, action_dim=3):
        super(Controller, self).__init__()
        self.model = nn.Sequential(nn.Linear(state_dim, action_dim))

    def forward(self, x):
        raw_actions = self.model(x)
        steering = torch.tanh(raw_actions[:, 0:1])
        gas = torch.sigmoid(raw_actions[:, 1:2])
        brake = torch.sigmoid(raw_actions[:, 2:3]) * 0.8
        brake = brake * (1 - gas)
        return torch.cat([steering, gas, brake], dim=1)

    def get_action(self, state):
        with torch.no_grad():
            action = self.forward(state)
        return action.squeeze()

# ==========================================
# 2. LOAD WEIGHTS CHO VAE VÀ RNN TỪ DRIVE
# ==========================================
print("📥 Đang nạp não bộ (VAE + RNN)...")
vision = VAE(3, LATENT_DIM).to(device)
vision.load_state_dict(torch.load(f'{DRIVE_PATH}/checkpoints/vae_10k_3epochs.pth', map_location=device, weights_only=True))
vision.eval()

memory = MDNRNN(latent_dim=LATENT_DIM, action_dim=ACTION_DIM, hidden_dim=HIDDEN_DIM, num_gaussians=NUM_GAUSSIANS).to(device)
memory.load_state_dict(torch.load(f'{DRIVE_PATH}/checkpoints/rnn_10k_final.pth', map_location=device, weights_only=True))
memory.eval()
print("✅ Nạp xong!")

# ==========================================
# 3. CÁC HÀM PHỤ TRỢ (ĐÃ FIX LỖI NHIỄU VAE)
# ==========================================
def make_env(name='CarRacing-v3'):
    def _init():
        return gym.make(name, render_mode='rgb_array', lap_complete_percent=1.0, domain_randomize=False, continuous=True)
    return _init

def create_vector_envs(num_envs):
    return AsyncVectorEnv([make_env() for _ in range(num_envs)], shared_memory=True)

def encode_obs_batch(obs_batch, size=(96, 96), device=device):
    obs_tensor = torch.as_tensor(obs_batch, dtype=torch.float32, device=device)
    obs_tensor = obs_tensor.permute(0, 3, 1, 2) / 255.0
    obs_tensor = obs_tensor[:, :, :-12, :]
    obs_tensor = F.interpolate(obs_tensor, size=size, mode='bicubic')
    with torch.no_grad():
        mu, _ = vision.encode(obs_tensor)
        # 🚨 ĐÃ FIX: Chỉ lấy mu, BỎ logvar (nhiễu) để mắt nhìn rõ ràng
        z = mu
    return z

def process_actions(controllers, x):
    return torch.stack([controllers[i].get_action(x[i:i+1]) for i in range(x.size(0))], dim=0)

def load_weights(controller_class, solutions):
    controllers = []
    with torch.no_grad():
        for params in solutions:
            ctrl = controller_class(state_dim=LATENT_DIM + HIDDEN_DIM, action_dim=ACTION_DIM).to(device)
            torch.nn.utils.vector_to_parameters(torch.tensor(params, dtype=torch.float32).to(device), ctrl.parameters())
            controllers.append(ctrl)
    return controllers

def evaluate_policies(solutions, controller_class, max_steps, memory):
    num_policies = len(solutions)
    controllers = load_weights(controller_class, solutions)
    envs = create_vector_envs(num_envs=num_policies)
    obs, _ = envs.reset()
    hidden = memory.rnn.init_hidden(num_policies, device)
    cumulative_rewards = np.zeros(num_policies)
    dones = np.full(num_policies, False)

    with torch.no_grad():
        for _ in range(max_steps):
            if np.all(dones): break
            z_batch = encode_obs_batch(obs)
            h = hidden[0].squeeze(0)
            x = torch.cat([z_batch, h], dim=-1)
            actions = process_actions(controllers, x)
            obs, rewards, dones_new, _, _ = envs.step(actions.cpu().numpy())

            z_batch = z_batch.unsqueeze(1)
            actions = actions.unsqueeze(1)
            _, hidden = memory.rnn(z_batch, actions, hidden)

            dones = np.logical_or(dones, dones_new)
            cumulative_rewards += rewards * (~dones)

    envs.close()
    return cumulative_rewards.tolist()

# ==========================================
# 4. THUẬT TOÁN CMA-ES BẢN CHUẨN (TRAIN LẠI TỪ ĐẦU)
# ==========================================
def train_cma_es_full_v2(controller_class, memory, max_generations=200, max_steps=1000, popsize=64, rollouts=5):
    os.makedirs(f'{DRIVE_PATH}/checkpoints/cma', exist_ok=True)
    metrics = {'generation': [], 'best_reward': [], 'mean_reward': [], 'worst_reward': []}

    controller = Controller(state_dim=LATENT_DIM + HIDDEN_DIM, action_dim=ACTION_DIM).to(device)
    initial_params = torch.nn.utils.parameters_to_vector(controller.parameters()).detach().cpu().numpy()
    es = cma.CMAEvolutionStrategy(initial_params, 0.1, {'popsize': popsize})

    print(f"🚀 Bắt đầu train CMA-ES bản V2 (Tắt nhiễu VAE, Rollouts={rollouts})...")
    for generation in range(1, max_generations + 1):
        start_time = time.time()
        solutions = es.ask()

        mean_rewards = []
        for _ in range(rollouts):
            rewards = evaluate_policies(solutions, controller_class, max_steps, memory)
            mean_rewards.append(rewards)
        mean_rewards = np.mean(mean_rewards, axis=0)

        es.tell(solutions, [-r for r in mean_rewards])

        best_rew, mean_rew, worst_rew = np.max(mean_rewards), np.mean(mean_rewards), np.min(mean_rewards)
        metrics['generation'].append(generation)
        metrics['best_reward'].append(best_rew)
        metrics['mean_reward'].append(mean_rew)
        metrics['worst_reward'].append(worst_rew)
        pd.DataFrame(metrics).to_csv(f"{DRIVE_PATH}/checkpoints/cma/cma_es_metrics_v2.csv", index=False)

        elapsed_time = time.time() - start_time
        print(f'Gen {generation}/{max_generations} | Best: {best_rew:.1f} | Avg: {mean_rew:.1f} | Worst: {worst_rew:.1f} | Time: {elapsed_time:.0f}s')

    best_controller = Controller(state_dim=LATENT_DIM + HIDDEN_DIM, action_dim=ACTION_DIM).to(device)
    torch.nn.utils.vector_to_parameters(torch.tensor(es.result.xbest, dtype=torch.float32).to(device), best_controller.parameters())
    torch.save(best_controller.state_dict(), f'{DRIVE_PATH}/checkpoints/controller_test_v2.pth')
    print("✅ Đã lưu Controller V2 xịn!")
    return best_controller

# 🔥 BẤM CHẠY CELL NÀY LÀ XONG 🔥
# Tăng rollouts lên 5 để ép agent thi 5 đường đua khác nhau mỗi thế hệ
best_controller = train_cma_es_full_v2(
    Controller, memory,
    max_generations=200,
    max_steps=1000,
    popsize=64,
    rollouts=5
)
```

```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML
import numpy as np
import torch
import gymnasium as gym

def render_best_of_n_colab(env_name, controller, mdnrnn, num_episodes=5, max_steps=1000):
    device = next(controller.parameters()).device
    env = gym.make(env_name, render_mode='rgb_array', lap_complete_percent=1.0, continuous=True)

    controller.eval()
    mdnrnn.eval()
    # vision.eval() # Đảm bảo VAE cũng ở chế độ eval

    best_reward = -float('inf')
    best_frames = []

    print(f"🏆 BẮT ĐẦU TỔ CHỨC GIẢI ĐUA {num_episodes} VÒNG (ĐÃ FIX LỖI ẢO GIÁC)...")

    with torch.no_grad():
        for ep in range(num_episodes):
            # BỎ SEED ĐỂ XE CHẠY ĐƯỜNG RANDOM GIỐNG LÚC TRAIN
            obs, _ = env.reset()
            h = mdnrnn.rnn.init_hidden(1, device)

            current_frames = [obs]
            cumulative_reward = 0

            for _ in range(max_steps):
                # 1. Mắt nhìn (ĐÃ SỬA: DÙNG MU THAY VÌ SAMPLE)
                obs_tensor = torch.as_tensor(obs[np.newaxis, ...], dtype=torch.float32, device=device)
                obs_tensor = obs_tensor.permute(0, 3, 1, 2) / 255.0
                obs_tensor = obs_tensor[:, :, :-12, :]
                obs_tensor = F.interpolate(obs_tensor, size=(96, 96), mode='bicubic')

                mu, _ = vision.encode(obs_tensor)
                z = mu # Lấy trung bình, bỏ qua nhiễu logvar!

                # 2. Não nghĩ
                x = torch.cat([z, h[0].squeeze(0)], dim=-1)
                a = controller.get_action(x)

                # 3. Hành động
                action_np = a.cpu().numpy()
                obs, reward, terminated, truncated, _ = env.step(action_np)
                done = terminated or truncated

                # 4. Cập nhật trí nhớ
                _, h = mdnrnn.rnn(z.unsqueeze(1), a.view(1, 1, -1), h=h)

                cumulative_reward += reward
                current_frames.append(obs)

                if done: break

            print(f"🏁 Vòng {ep + 1}/{num_episodes} | Điểm số: {cumulative_reward:.2f}")

            if cumulative_reward > best_reward:
                best_reward = cumulative_reward
                best_frames = current_frames.copy()

    env.close()
    print(f"\n🌟 QUÁN QUÂN SAU KHI FIX LỖI ĐẠT: {best_reward:.2f} điểm.")
    print(f"🎞️ Đang render video {len(best_frames)} frames của nhà vô địch...")

    # Khâu xuất Video HTML5 (Giữ nguyên)
    fig, ax = plt.subplots(figsize=(6, 6))
    fig.subplots_adjust(left=0, bottom=0, right=1, top=1, wspace=None, hspace=None)
    ax.axis('off')
    im = ax.imshow(best_frames[0])

    def update(i):
        im.set_array(best_frames[i])
        return [im]

    anim = animation.FuncAnimation(fig, update, frames=len(best_frames), interval=50, blit=True)
    plt.close()

    return HTML(anim.to_jshtml())

# Cho xe chạy 5 vòng liên tiếp, chỉ lấy video của vòng đỉnh nhất!
render_best_of_n_colab('CarRacing-v3', best_controller, memory, num_episodes=32, max_steps=1000)
```
